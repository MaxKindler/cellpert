```python
import pandas as pd
import numpy as np
```

```python
df_baseline_seen_unseen_standard = pd.read_csv("data/baseline_results_N-DE.csv")
df_baseline_unseen_sub = pd.read_csv("data/results/final/baseline_results_unseen_subsample.csv")
```

```python
metric_cols = ["DE_F1", "DE_Precision", "DE_Recall", "Energy_Loss", "MSE"]
```

```python
df_baseline_seen_standard = df_baseline_seen_unseen_standard[df_baseline_seen_unseen_standard["task"] == "SEEN"]
df_baseline_unseen_standard = df_baseline_seen_unseen_standard[df_baseline_seen_unseen_standard["task"] == "UNSEEN-PERT"]
```

```python
df_baseline_seen_standard_mean = df_baseline_seen_standard[df_baseline_seen_standard["baseline"] == "mean"]
df_baseline_unseen_standard_mean = df_baseline_unseen_standard[df_baseline_unseen_standard["baseline"] == "mean"]
df_baseline_unseen_sub_mean = df_baseline_unseen_sub[df_baseline_unseen_sub["baseline"] == "mean"]

df_baseline_seen_standard_mean_shift = df_baseline_seen_standard[df_baseline_seen_standard["baseline"] == "mean_shift"]
df_baseline_unseen_standard_mean_shift = df_baseline_unseen_standard[df_baseline_unseen_standard["baseline"] == "mean_shift"]
df_baseline_unseen_sub_mean_shift = df_baseline_unseen_sub[df_baseline_unseen_sub["baseline"] == "mean_shift"]
```
