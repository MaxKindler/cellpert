```python
import os
os.environ["CUDA_VISIBLE_DEVICES"]= '0'

from scipy import stats
import plotly.express as px
import plotly.graph_objects as go
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
import ast
```

```python
# Load data
df_transformer = pd.read_csv("data/results/final/wandb_transformer_seen_unseen.csv")
df_rf = pd.read_csv("data/results/final/wandb_rectified_flow_seen_unseen.csv")
df_tf = pd.read_csv("data/results/final/wandb_transformerflow_seen_unseen.csv")
df_baselines = pd.read_csv("data/results/final/baseline_results.csv")
```

```python
cols_of_interest = ["Name", "seed", "model_family", "task"]
val_metric_cols = ["de_val_score", "de_precision_val_score", "de_recall_val_score", "energy_val_score", "mse_val_loss"]
test_metric_cols = ["de_test_score", "de_precision_test_score", "de_recall_test_score", "energy_test_score", "mse_test_loss"]

baseline_cols_of_interest = ["task", "baseline", "seed"]
baseline_cols = ["DE_F1", "DE_Precision", "DE_Recall", "Energy_Loss", "MSE"]
baseline_cols_all = ["DE-F1_all", "DE-Precision_all", "DE-Recall_all", "Energy_Loss_all", "MSE_all"]

# mean_baseline_cols = ["mean_baseline_de_score", "mean_baseline_precision_score", "mean_baseline_recall_score", "mean_baseline_energy_score", "mean_baseline_mse_score"]
# mean_shift_baseline_cols = ["mean_shift_baseline_de_score", "mean_shift_baseline_precision_score", "mean_shift_baseline_recall_score", "mean_shift_baseline_energy_score", "mean_shift_baseline_mse_score"]

all_values_cols = ["de_test_score_all", "de_precition_test_score_all", "de_recall_test_score_all", "energy_test_score_all", "mse_test_loss_all"]
```

```python
df_transformer = df_transformer[cols_of_interest + val_metric_cols + test_metric_cols + all_values_cols]
df_rf = df_rf[cols_of_interest + val_metric_cols + test_metric_cols + all_values_cols]
df_tf = df_tf[cols_of_interest + val_metric_cols + test_metric_cols + all_values_cols]
df_baselines = df_baselines[baseline_cols_of_interest + baseline_cols + baseline_cols_all]
```

```python
df_mean_baseline = df_baselines[df_baselines["baseline"] == "mean"]
df_mean_shift_baseline = df_baselines[df_baselines["baseline"] == "mean_shift"]
```

```python
df_transformer_seen = df_transformer[df_transformer["task"] == "SEEN"]
df_transformer_unseen = df_transformer[df_transformer["task"] == "UNSEEN-PERT"]

df_rf_seen = df_rf[df_rf["task"] == "SEEN"]
df_rf_unseen = df_rf[df_rf["task"] == "UNSEEN-PERT"]

df_tf_seen = df_tf[df_tf["task"] == "SEEN"]
df_tf_unseen = df_tf[df_tf["task"] == "UNSEEN-PERT"]

df_mean_baseline_seen = df_mean_baseline[df_mean_baseline["task"] == "SEEN"]
df_mean_baseline_unseen = df_mean_baseline[df_mean_baseline["task"] == "UNSEEN-PERT"]

df_mean_shift_baseline_seen = df_mean_shift_baseline[df_mean_shift_baseline["task"] == "SEEN"]
df_mean_shift_baseline_unseen = df_mean_shift_baseline[df_mean_shift_baseline["task"] == "UNSEEN-PERT"]
```

```python
df_transformer_seen.sort_values(by=["seed"], inplace=True)
df_transformer_unseen.sort_values(by=["seed"], inplace=True)

df_rf_seen.sort_values(by=["seed"], inplace=True)
df_rf_unseen.sort_values(by=["seed"], inplace=True)

df_tf_seen.sort_values(by=["seed"], inplace=True)
df_tf_unseen.sort_values(by=["seed"], inplace=True)

df_mean_baseline_seen.sort_values(by=["seed"], inplace=True)
df_mean_baseline_unseen.sort_values(by=["seed"], inplace=True)
df_mean_shift_baseline_seen.sort_values(by=["seed"], inplace=True)
df_mean_shift_baseline_unseen.sort_values(by=["seed"], inplace=True)
```

# Holm-corrected Statistical Tests
Paired runs are aligned on `seed` (the seed determines the data split). Within a task, models and baselines are compared with the two-sided Wilcoxon signed-rank test. Seen vs. unseen uses the two-sided Mann-Whitney U (rank-sum) test, since the runs are not paired.

Holm correction is reported for two families:
- `p_holm_all`: one family over all tests
- `p_holm_task_metric`: separate family per task x metric

```python
def holm(pvals):
    """Holm-Bonferroni adjusted p-values."""
    p = np.asarray(pvals, dtype=float)
    m = len(p)
    order = np.argsort(p)
    adj = np.empty(m)
    running_max = 0.0
    for rank, idx in enumerate(order):
        running_max = max(running_max, (m - rank) * p[idx])
        adj[idx] = min(running_max, 1.0)
    return adj


def to_metrics(df, cols):
    # rename baseline columns to the model metric names so all dataframes share one schema
    return df[["seed"] + cols].rename(columns=dict(zip(cols, test_metric_cols)))


def compare_paired(task, name_a, df_a, cols_a, name_b, df_b, cols_b):
    merged = to_metrics(df_a, cols_a).merge(to_metrics(df_b, cols_b), on="seed", suffixes=("_a", "_b"))
    assert len(merged) == 20, f"{task} {name_a} vs. {name_b}: only {len(merged)} matched seeds"
    rows = []
    for metric in test_metric_cols:
        a, b = merged[metric + "_a"], merged[metric + "_b"]
        w, p = stats.wilcoxon(a, b)
        rows.append({"task": task, "comparison": f"{name_a} vs. {name_b}", "metric": metric,
                     "median_a": a.median(), "median_b": b.median(), "W": w, "p": p})
    return rows
```

```python
paired_results = []

for task, models, base_mean, base_shift in [
    ("SEEN", {"Transformer": df_transformer_seen, "Rectified Flow": df_rf_seen, "Transformer Flow": df_tf_seen},
     df_mean_baseline_seen, df_mean_shift_baseline_seen),
    ("UNSEEN", {"Transformer": df_transformer_unseen, "Rectified Flow": df_rf_unseen, "Transformer Flow": df_tf_unseen},
     df_mean_baseline_unseen, df_mean_shift_baseline_unseen),
]:
    names = list(models)
    for i in range(len(names)):
        for j in range(i + 1, len(names)):
            paired_results += compare_paired(task, names[i], models[names[i]], test_metric_cols,
                                             names[j], models[names[j]], test_metric_cols)
    for n in names:
        paired_results += compare_paired(task, n, models[n], test_metric_cols, "Mean Baseline", base_mean, baseline_cols)
        paired_results += compare_paired(task, n, models[n], test_metric_cols, "Mean Shift Baseline", base_shift, baseline_cols)

res_paired = pd.DataFrame(paired_results)
res_paired["p_holm_all"] = holm(res_paired["p"])
res_paired["p_holm_task_metric"] = res_paired.groupby(["task", "metric"])["p"].transform(holm)
res_paired["significant_all"] = res_paired["p_holm_all"] < 0.05
res_paired["significant_task_metric"] = res_paired["p_holm_task_metric"] < 0.05
```

```python
pd.set_option("display.max_rows", 200)
np.round(res_paired, 4)
```

## Seen vs. Unseen (Mann-Whitney U, Holm-corrected)
Seen and unseen runs use different test sets, so they are treated as independent samples. `p_holm_metric` corrects within each metric (the analogue of task x metric here).

```python
seen_unseen_results = []

for name, df_seen, df_unseen, cols in [
    ("Transformer", df_transformer_seen, df_transformer_unseen, test_metric_cols),
    ("Rectified Flow", df_rf_seen, df_rf_unseen, test_metric_cols),
    ("Transformer Flow", df_tf_seen, df_tf_unseen, test_metric_cols),
    ("Mean Baseline", df_mean_baseline_seen, df_mean_baseline_unseen, baseline_cols),
    ("Mean Shift Baseline", df_mean_shift_baseline_seen, df_mean_shift_baseline_unseen, baseline_cols),
]:
    seen, unseen = to_metrics(df_seen, cols), to_metrics(df_unseen, cols)
    for metric in test_metric_cols:
        u, p = stats.mannwhitneyu(seen[metric], unseen[metric], alternative="two-sided")
        seen_unseen_results.append({"model": name, "metric": metric, "median_seen": seen[metric].median(),
                                    "median_unseen": unseen[metric].median(), "U": u, "p": p})

res_seen_unseen = pd.DataFrame(seen_unseen_results)
res_seen_unseen["p_holm_all"] = holm(res_seen_unseen["p"])
res_seen_unseen["p_holm_metric"] = res_seen_unseen.groupby("metric")["p"].transform(holm)
res_seen_unseen["significant_all"] = res_seen_unseen["p_holm_all"] < 0.05
res_seen_unseen["significant_metric"] = res_seen_unseen["p_holm_metric"] < 0.05
np.round(res_seen_unseen, 4)
```
